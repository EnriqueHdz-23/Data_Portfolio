1. Un cliente llega a un mostrador cada 15 minutos en promedio. Si podemos modelar la llegada de clientes mediante
una **distribución exponencial**, ¿cuál es la probabilidad de esperar menos de 7 minutos entre clientes?

```python
from scipy.stats import expon

# Lambda (tasa de llegada) es el inverso del tiempo promedio entre llegadas.
# Tiempo promedio entre llegadas = 15 minutos
# Lambda = 1 / 15 minutos = 1/15 clientes por minuto

lambd = 1/15

# Calculamos la probabilidad de esperar menos de 7 minutos
probability = expon.cdf(7, scale=1/lambd)

print(f"La probabilidad de esperar menos de 7 minutos entre clientes es: {probability}")
```

2. Considerando las tablas `employees`, `departments` y `regions`. Se utiliza sintaxis SQL para resolver lo siguiente:
- a) Obtener el número de empleados que hay por departamento pero considerando sólo los departamentos de 'Computers', 'Beauty' y 'Games' 
- b) Crear un query que usando joins entre las tablas devuelva los campos 'first_name', 'last_name', 'division','region' y 'country'.

```python
#pip install ipython-sql
#!pip install pandasql
```

```python
from pandasql import sqldf

query = """
SELECT department, COUNT(*) AS num_employees
FROM employees
WHERE department IN ('Computers', 'Beauty', 'Games')
GROUP BY department;
"""

result = sqldf(query, locals())
print(result)
```

```python
query2 = """
SELECT e.first_name, e.last_name, d.division, r.region, r.country
FROM employees e
JOIN departments d ON e.department = d.department
JOIN regions r ON e.region_id = r.region_id;
"""
result2 = sqldf(query2, locals())
print(result2)
```
