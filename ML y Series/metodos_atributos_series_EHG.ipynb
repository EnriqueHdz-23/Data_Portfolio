## Ejemplo: Machine Learning con Pandas

```python
df
```

```python
iris_forma2
```

```python
iris_forma2['target']
```

```python
df['target']=iris_forma2['target'] # agregar una columna
df
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
# Nos ayuda a distinguir las categorías, es un árbol de decisión en su modelo de clasificación
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test, Y_train,Y_test= train_test_split(df[['sepal length (cm)','sepal width (cm)','petal length (cm)','petal width (cm)']],
                                                 df['target'], random_state=0)
```

```python
# 75% de las filas las toma en Training
# 25% de las filas las toma en Test
```

```python
X_train
```

```python
Y_train
```

```python
X_test
```

```python
clf=DecisionTreeClassifier(max_depth=2,
                           random_state=0)
clf.fit(X_train,Y_train)
```

```python
# para Graficar el árbol

fn=['sepal length (cm)','sepal width (cm)','petal length (cm)','petal width (cm)']
cn=['setosa','versicolor','virginica']
fig,axes=plt.subplots(nrows=1,ncols=1,figsize=(4,4),dpi=300)
_=tree.plot_tree(clf,
                feature_names=fn,
                class_names=cn,
                filled=True)
```

```python
# indice de "gini" nos dice que tan distribuida esta la base (ginis muy altos nos dice que hay una distribución más uniforme )
# gini = 0 no hay ganancia de información
```

```python
# en el árbol de decisión los resultados de la izquierda son SI y la derecha es NO en cuanto a cumplir la condición
```

```python
# Cada nodo muestra:
# Condición (ej. petal length (cm) <= 4.95)
# gini: impureza del nodo (0 = puro)
# samples: número de muestras en ese nodo
# value: distribución de clases
# class: clase mayoritaria
```
