4. El `factorial` de un entero no negativo **n**, denotado por **n!**,  se define como el producto *n!* =  *1x 2 x 3 x...x (n-1) x n*. Por ejemplo: *4! = 1 x 2 x 3 x 4 = 24*.  Se escribe un programa que calcule el `factorial` de un número entre 0 y 30.

```python
def factorial(n):
  if n < 0 or n > 30:
    raise ValueError("El número debe estar entre 0 y 30")
  if n == 0:
    return 1
  else:
    return reduce(lambda x, y: x * y, range(1, n + 1))
```

```python
print("Escribe un número entre 0 y 30: ")
n = int(input())
print(f"El factorial de {n} es: {factorial(n)}")
```

8.  Calculando la media (promedio) de cada una de las variables 'Avg_User_IMDB'  y 'Avg_User_Rtn_Tom' y guarda estos valores en mean_1 y mean_2, respectivamente.

    **a)** Escribiendo una función llamada `califica( )` que reciba como argumento un `renglón` y califique a la película de  la siguiente manera:
    
    Si el actor es ` 'Roger Moore'`  o `'Daniel Craig'`   y además se cumple que  `Avg_User_IMDB > mean_1 `      Y  `Avg_User_Rtn_Tom > mean_2` , entonces regrese 'excelente' , de lo contrario que regrese 'buena'.

    **b)** Utilizando el método `apply( )`  para aplicar la función `califica()`  a cada uno de los renglones del data set.
    

```python
mean_1 = df['Avg_User_IMDB'].mean()
mean_2 = df['Avg_User_Rtn_Tom'].mean()

# a)
def califica(row):
    if (row['Bond Actor'] == 'Roger Moore' or row['Bond Actor'] == 'Daniel Wroughton Craig') and row['Avg_User_IMDB'] > mean_1 and row['Avg_User_Rtn_Tom'] > mean_2:
        return 'excelente'
    else:
        return 'buena'

# b)
df['Calificacion'] = df.apply(califica, axis=1)
```
