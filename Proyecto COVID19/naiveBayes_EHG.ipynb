```python
from sklearn.naive_bayes import BernoulliNB
from sklearn.naive_bayes import MultinomialNB
from sklearn.naive_bayes import GaussianNB
import pandas as pd
import numpy as np

df = pd.read_csv('datos_abiertos_covid19.csv',nrows=6000)
df = df.sample(frac=1)
df = df.head(6000)
```

```python
df = df.dropna(subset=['INTUBADO'])
```

```python
Y = df['INTUBADO'] == 1
```

```python
df['SEXO'] = df['SEXO'].apply(str)
df['DIABETES'] = df['DIABETES'].apply(str)
df['TABAQUISMO'] = df['TABAQUISMO'].apply(str)
df['OBESIDAD'] = df['OBESIDAD'].apply(str)
df['NEUMONIA'] = df['NEUMONIA'].apply(str)
df['HIPERTENSION'] = df['HIPERTENSION'].apply(str)

X = pd.get_dummies(data=df[['SEXO','DIABETES','TABAQUISMO','OBESIDAD','NEUMONIA','HIPERTENSION','EDAD']])
```

```python
X.head()
```

```python
clf = BernoulliNB()
#clf = MultinomialNB()
clf.fit(X,Y)
Y_pred = clf.predict(X)
```

```python
np.mean(Y == Y_pred)
```
