```python
#Test de la Chi-Cuadrado

import pandas as pd
import numpy as np 
df = pd.read_csv('datos_abiertos_covid19.csv',nrows=1000)
```

```python
np.random.seed(0)
df = df[df['ENTIDAD_UM'].isin(['25','27'])]
df = df.sample(frac=1)
df = df[0:800] 
```

```python
df['Suceptible'] = df['CARDIOVASCULAR'] == 2
observados = pd.crosstab(index=df['Suceptible'],columns=df['ENTIDAD_UM'], margins=True)
```

```python
observados
```

```python
from scipy.stats import chi2_contingency
```

```python
test = chi2_contingency(observados)
```

```python
test
```

```python
esperados = pd.DataFrame(test[3])
```

```python
esperados
```

```python
esperados_rel = round(esperados.apply(lambda r: r/len(df) *100,axis=1),2)
```

```python
observados_rel = round(observados.apply(lambda r: r/len(df) *100,axis=1),2)
```
