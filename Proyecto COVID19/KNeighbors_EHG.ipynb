```python
import numpy as np
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier


df = pd.read_csv('datos_abiertos_covid19.csv',nrows=6000)
```

```python
newdf = df[['EDAD','OBESIDAD','EPOC','TABAQUISMO','OBESIDAD','INTUBADO']].dropna()
cols = newdf[newdf.columns[newdf.columns != 'INTUBADO']]
```

```python
filtro = newdf['INTUBADO'] == 1
```

```python
newdf['INTUBADO'][filtro] = 'Intubado'
newdf['INTUBADO'][filtro == False] = 'No Intubado'
```

```python
newdf['INTUBADO'].head()
```

```python
nbrs_3 = KNeighborsClassifier(n_neighbors = 3,n_jobs = -1)
```

```python
nbrs_3.fit(cols,newdf['INTUBADO'])
```

```python
predicciones_3 = nbrs_3.predict(cols)
```

```python
np.mean(predicciones_3 == newdf['INTUBADO'])
```

```python
np.mean(newdf['INTUBADO'] == 'No Intubado')
```

```python
nbrs_1 = KNeighborsClassifier(n_neighbors = 1,n_jobs = -1)
nbrs_1.fit(cols,newdf['INTUBADO'])
predicciones_1 = nbrs_1.predict(cols)
np.mean(predicciones_1 == newdf['INTUBADO'])
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix = confusion_matrix(newdf['INTUBADO'],predicciones_1)
print(confusion_matrix)
```
