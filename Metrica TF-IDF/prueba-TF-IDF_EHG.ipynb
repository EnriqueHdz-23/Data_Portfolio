```python
# Creando un ejemplo corto y sencillo para usar la métrica TF-IDF

import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

# Ejemplo de documentos
documents = [
    "Este es el primer documento sobre procesamiento del lenguaje.",
    "El segundo documento también trata sobre lenguaje.",
    "Este es el tercer documento, pero habla de otro tema.",
]

# Crear un vectorizador TF-IDF
tfidf_vectorizer = TfidfVectorizer()

# Ajustar y transformar los documentos
tfidf_matrix = tfidf_vectorizer.fit_transform(documents)

# Convertir la matriz TF-IDF a un DataFrame de pandas
df = pd.DataFrame(tfidf_matrix.toarray(), columns=tfidf_vectorizer.get_feature_names_out())

# Mostrar el DataFrame
df


#Explicación:

#La palabra "documento" aparece en todos los documentos, por lo tanto tiene un bajo valor TF-IDF.
#La palabra "primer" aparece solo en el primer documento, por lo tanto tiene un alto valor TF-IDF en ese documento.
#Las palabras que aparecen solo en algunos documentos tienen un valor TF-IDF más alto en esos documentos, mientras que las que
# aparecen en muchos documentos tienen un menor valor TF-IDF.

```

```python
# TF-IDF with Naive Bayes Classification

import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report

# 1. Sample dataset (synthetic for demonstration)
data = {
    "text": [
        "I love this product, it is amazing!",
        "Terrible experience, I will not buy again.",
        "The quality is decent, nothing special.",
        "Absolutely fantastic service and great value.",
        "Worst purchase ever, very disappointed.",
        "Good price but delivery was slow.",
        "Excellent! Highly recommend to everyone.",
        "Not worth the money, poor quality."
    ],
    "label": ["positive", "negative", "neutral", "positive", "negative", "neutral", "positive", "negative"]
}

df = pd.DataFrame(data)

# 2. TF-IDF Vectorization
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(df["text"])
y = df["label"]

# 3. Train-test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# 4. Naive Bayes Classifier
model = MultinomialNB()
model.fit(X_train, y_train)

# 5. Predictions
y_pred = model.predict(X_test)

# 6. Evaluation
print("Classification Report:\n")
print(classification_report(y_test, y_pred))

# 7. Example prediction
sample_texts = ["I am very happy with the service", "This was a waste of money"]
sample_features = vectorizer.transform(sample_texts)
predictions = model.predict(sample_features)

for text, pred in zip(sample_texts, predictions):
    print(f"Text: '{text}' --> Predicted Sentiment: {pred}")
```
